## Setup

```python
from database import db
from analytics import analytics
import pandas as pd

# Connect to database
if not db.connection:
    db.connect()

print("✅ Ready to run analytics!")
```

## Insight 1: Best Performing ETFs

```python
# Get best performing ETFs
results = analytics.get_best_performing_etfs(limit=15)

if results:
    df = pd.DataFrame(results)
    print("📊 TOP 15 BEST PERFORMING ETFs\n")
    display(df[['ticker_symbol', 'etf_name', 'asset_class', 'total_return_pct']])
    
    # Visualization
    import matplotlib.pyplot as plt
    plt.figure(figsize=(12, 6))
    plt.barh(df['ticker_symbol'][:10], df['total_return_pct'][:10])
    plt.xlabel('Total Return (%)')
    plt.title('Top 10 Best Performing ETFs')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()
else:
    print("❌ No data available")
```

## Insight 2: Volatility Analysis

```python
# Analyze volatility for all ETFs
results = analytics.get_etf_volatility(limit=15)

if results:
    df = pd.DataFrame(results)
    print("📊 VOLATILITY ANALYSIS (Top 15 Most Volatile)\n")
    display(df[['ticker_symbol', 'etf_name', 'annualized_volatility_pct', 'annualized_return_pct']])
    
    # Risk-Return Scatter
    import matplotlib.pyplot as plt
    plt.figure(figsize=(10, 6))
    plt.scatter(df['annualized_volatility_pct'], df['annualized_return_pct'])
    for idx, row in df.head(10).iterrows():
        plt.annotate(row['ticker_symbol'], 
                    (row['annualized_volatility_pct'], row['annualized_return_pct']))
    plt.xlabel('Volatility (%)')
    plt.ylabel('Return (%)')
    plt.title('Risk vs Return')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
else:
    print("❌ No data available")
```

## Insight 3: Correlation Analysis

```python
# Find highly correlated ETF pairs
results = analytics.get_correlation_pairs(min_correlation=0.7)

if results:
    df = pd.DataFrame(results)
    print("📊 HIGHLY CORRELATED ETF PAIRS (correlation > 0.7)\n")
    display(df.head(20))
    
    print(f"\nTotal pairs found: {len(df)}")
    print(f"Highest correlation: {df['correlation'].max():.4f}")
    print(f"Lowest correlation: {df['correlation'].min():.4f}")
else:
    print("❌ No highly correlated pairs found")
```

## Insight 4: Maximum Drawdown

```python
# Analyze maximum drawdown for each ETF
results = analytics.get_max_drawdown_by_etf(limit=15)

if results:
    df = pd.DataFrame(results)
    print("📊 MAXIMUM DRAWDOWN ANALYSIS (Top 15 Worst)\n")
    display(df[['ticker_symbol', 'etf_name', 'asset_class', 'max_drawdown_pct']])
    
    # Visualization
    import matplotlib.pyplot as plt
    plt.figure(figsize=(12, 6))
    plt.barh(df['ticker_symbol'][:10], df['max_drawdown_pct'][:10], color='red')
    plt.xlabel('Maximum Drawdown (%)')
    plt.title('Top 10 Worst Drawdowns')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()
else:
    print("❌ No data available")
```

## Insight 5: Asset Class Performance

```python
# Compare performance across asset classes
results = analytics.get_asset_class_performance()

if results:
    df = pd.DataFrame(results)
    print("📊 ASSET CLASS PERFORMANCE\n")
    display(df)
    
    # Visualization
    import matplotlib.pyplot as plt
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Average returns
    ax1.barh(df['asset_class'], df['avg_return_pct'])
    ax1.set_xlabel('Average Return (%)')
    ax1.set_title('Average Returns by Asset Class')
    ax1.grid(True, alpha=0.3)
    
    # Number of ETFs
    ax2.pie(df['num_etfs'], labels=df['asset_class'], autopct='%1.0f%%')
    ax2.set_title('ETF Distribution by Asset Class')
    
    plt.tight_layout()
    plt.show()
else:
    print("❌ No data available")
```

## Insight 8: Expense Ratio Impact

```python
# Analyze correlation between expense ratio and performance
results = analytics.analyze_expense_ratio_impact()

if results:
    df = pd.DataFrame(results)
    print("📊 EXPENSE RATIO IMPACT ANALYSIS\n")
    display(df)
    
    # Visualization
    import matplotlib.pyplot as plt
    plt.figure(figsize=(10, 6))
    plt.bar(df['expense_category'], df['avg_return_pct'])
    plt.xlabel('Expense Category')
    plt.ylabel('Average Return (%)')
    plt.title('Impact of Expense Ratio on Returns')
    plt.xticks(rotation=45, ha='right')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
else:
    print("❌ No data available")
```

## Insight 9: Concentration Risk

```python
# Analyze portfolio concentration using HHI
results = analytics.analyze_concentration_risk()

if results:
    df = pd.DataFrame(results)
    print("📊 CONCENTRATION RISK ANALYSIS (HHI)\n")
    display(df.head(15))
    
    print("\n📈 Distribution by Concentration Level:")
    print(df['concentration_level'].value_counts())
    
    # Visualization
    import matplotlib.pyplot as plt
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # HHI by portfolio
    top10 = df.head(10)
    ax1.barh(top10['benchmark_name'], top10['hhi_index'], color='orange')
    ax1.set_xlabel('HHI Index')
    ax1.set_title('Top 10 Most Concentrated Portfolios')
    ax1.axvline(x=0.25, color='r', linestyle='--', label='High Risk Threshold')
    ax1.legend()
    ax1.invert_yaxis()
    
    # Concentration levels
    conc_counts = df['concentration_level'].value_counts()
    ax2.pie(conc_counts.values, labels=conc_counts.index, autopct='%1.0f%%')
    ax2.set_title('Concentration Level Distribution')
    
    plt.tight_layout()
    plt.show()
else:
    print("❌ No data available")
```

## Insight 11: Portfolio vs Benchmark Comparison (UNIQUE FEATURE!)

```python
# Compare portfolio performance vs benchmark
scenario_id = 1  # Change to your scenario_id

comparison = analytics.compare_portfolio_vs_benchmark(scenario_id)

if comparison:
    print("📊 PORTFOLIO VS BENCHMARK COMPARISON\n")
    print(f"Scenario: {comparison['scenario_name']}")
    print(f"Benchmark: {comparison['benchmark_name'] or 'None'}\n")
    
    # Performance metrics
    metrics = {
        'Portfolio Return': comparison['portfolio_return'],
        'Portfolio Ann. Return': comparison['portfolio_ann_return'],
        'Portfolio Volatility': comparison['portfolio_volatility'],
        'Portfolio Sharpe': comparison['portfolio_sharpe'],
        'Portfolio Max DD': comparison['portfolio_max_dd'],
        'Alpha vs Benchmark': comparison['alpha'],
        'Performance': comparison['performance_vs_benchmark']
    }
    
    for key, value in metrics.items():
        print(f"   {key:25s}: {value}")
    
    # Holdings comparison
    print("\n📈 Holdings Comparison:")
    holdings_comp = analytics.get_benchmark_holdings_comparison(scenario_id)
    if holdings_comp:
        df = pd.DataFrame(holdings_comp)
        display(df.head(10))
else:
    print("ℹ️  Create and backtest a scenario with benchmark first")
```
