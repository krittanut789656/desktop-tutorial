```python
# ============================================================================
# ETF PORTFOLIO BACKTESTER - MINI LIVE DEMO (AUTOMATED)
# ONE-CELL EXECUTION: Press Shift+Enter and watch the complete demo!
# Duration: < 2 minutes
# ============================================================================

import sys
import os
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
from IPython.display import display, HTML
import warnings
warnings.filterwarnings('ignore')

# Import sub-modules
from modules.db_connector import DatabaseConnector, DatabaseConfig
from modules.backtest_engine import MomentumBacktester
from modules.analytics import PortfolioAnalytics
from modules.crud_operations import CRUDOperations

# Configure display
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 50)
%matplotlib inline
plt.style.use('seaborn-v0_8-darkgrid')

print("="*80)
print("🎬 ETF PORTFOLIO BACKTESTER - MINI LIVE DEMO")
print("="*80)
print("\n⏱️  Estimated Duration: < 2 minutes")
print("📊 Demonstrating: Complete Integrated System\n")

# ========== 1. SYSTEM INITIALIZATION ==========
print("\n" + "="*80)
print("[1/5] 🔧 SYSTEM INITIALIZATION")
print("="*80)
print("\nInitializing Main Controller and Sub-modules...")

# Configure database
config = DatabaseConfig()
config.host = '127.0.0.1'
config.port = 3306
config.user = 'root'
config.password = 'krittanut123456'  # ← Update if needed
config.database = 'etf_backtester_db'

# Initialize sub-modules
db = DatabaseConnector(config)
if not db.test_connection():
    print("\n❌ Database connection failed!")
    raise Exception("Cannot connect to database")

backtester = MomentumBacktester(db)
analytics = PortfolioAnalytics(db)
crud = CRUDOperations(db)

# Check data
etf_count = db.get_table_count('ETF_Master')
price_count = db.get_table_count('Price_Data')

print(f"\n✅ System Ready!")
print(f"   📊 {etf_count} ETFs | 📈 {price_count:,} price records")

# ========== 2. RUN QUICK BACKTEST ==========
print("\n" + "="*80)
print("[2/5] 🎯 RUNNING MOMENTUM BACKTEST")
print("="*80)
print("\nConfiguration: 90-day period | 90-day lookback | Top 5 ETFs")

# Quick backtest (3 months)
end_date = datetime.now()
start_date = end_date - timedelta(days=90)

results = backtester.run_backtest(
    start_date=start_date.strftime('%Y-%m-%d'),
    end_date=end_date.strftime('%Y-%m-%d'),
    lookback_days=90,
    holding_period_days=30,
    rebalance_days=30,
    top_n=5
)

print(f"\n📊 Backtest Summary:")
print(f"   Run ID: {results['run_id']}")
print(f"   Rebalances: {results['total_rebalances']}")
print(f"   Cumulative Return: {results['cumulative_return']*100:.2f}%")
print(f"   CAGR: {results['cagr']*100:.2f}%")

# ========== 3. GENERATE ANALYTICS INSIGHTS ==========
print("\n" + "="*80)
print("[3/5] 📊 GENERATING SQL-BASED ANALYTICS")
print("="*80)

# Insight 1: Volatility Analysis
print("\n🔍 Insight #1: Volatility Analysis by Asset Type")
vol_results = analytics.analyze_volatility_by_asset_type()

if vol_results:
    df_vol = pd.DataFrame(vol_results)
    print("\n📈 Volatility by Asset Type:")
    display(df_vol[['Asset_Type', 'Num_ETFs', 'Annualized_Volatility_Pct']].head())

# Insight 2: Lookback Comparison
print("\n🔍 Insight #2: Lookback Period Optimization")
lookback_results = analytics.compare_lookback_periods([90, 180])

if lookback_results:
    df_lookback = pd.DataFrame(lookback_results)
    print("\n📊 Lookback Period Performance:")
    display(df_lookback[['Lookback_Period_Days', 'Num_Runs', 'CAGR_Pct']].head())

# Insight 3: Drawdown Analysis
print("\n🔍 Insight #3: Drawdown Exposure Analysis")
drawdown_results = analytics.analyze_drawdown_exposure()

if drawdown_results:
    df_drawdown = pd.DataFrame(drawdown_results)
    print("\n⚠️  Asset Exposure During Drawdowns:")
    display(df_drawdown[['Asset_Type', 'Num_Occurrences', 'Avg_Weight_Pct']].head())

# ========== 4. CRUD OPERATIONS ==========
print("\n" + "="*80)
print("[4/5] 🔍 CRUD OPERATIONS DEMONSTRATION")
print("="*80)

print("\n📋 Reading Latest Backtest Results...")
crud_results = crud.read_backtest_results(limit=10)

if crud_results:
    df_crud = pd.DataFrame(crud_results)
    print(f"\n✅ Latest Backtest: {df_crud['Backtest_Run_ID'].iloc[0]}")
    print(f"   📅 Date: {df_crud['Run_Date'].iloc[0]}")
    print(f"   📊 Records: {len(df_crud)}")
    print(f"   📈 Avg Momentum: {df_crud['Momentum_Score'].mean():.2f}%")

# ========== 5. VISUALIZATIONS ==========
print("\n" + "="*80)
print("[5/5] 📊 GENERATING VISUALIZATIONS")
print("="*80)

# Create figure with subplots
fig = plt.figure(figsize=(18, 10))

# Plot 1: Volatility by Asset Type
if vol_results:
    ax1 = plt.subplot(2, 3, 1)
    df_vol_plot = pd.DataFrame(vol_results)
    ax1.bar(df_vol_plot['Asset_Type'], df_vol_plot['Annualized_Volatility_Pct'],
           color=['#e74c3c', '#3498db', '#2ecc71', '#f39c12'])
    ax1.set_title('Volatility by Asset Type', fontsize=12, fontweight='bold')
    ax1.set_ylabel('Volatility (%)')
    ax1.grid(axis='y', alpha=0.3)
    plt.xticks(rotation=45)

# Plot 2: Cumulative Returns (Major ETFs)
ax2 = plt.subplot(2, 3, 2)
query = """
SELECT em.Ticker_Symbol, pd.Price_Date, pd.Close_Price
FROM Price_Data pd
JOIN ETF_Master em ON pd.ETF_ID = em.ETF_ID
WHERE pd.Price_Date >= DATE_SUB(CURDATE(), INTERVAL 6 MONTH)
  AND em.Ticker_Symbol IN ('SPY', 'QQQ', 'AGG', 'GLD')
ORDER BY em.Ticker_Symbol, pd.Price_Date
"""
price_data = db.execute_query_dict(query)
if price_data:
    df_prices = pd.DataFrame(price_data)
    pivot = df_prices.pivot_table(index='Price_Date', columns='Ticker_Symbol', values='Close_Price')
    returns = pivot.pct_change()
    cum_returns = (1 + returns).cumprod()
    
    for col in cum_returns.columns:
        ax2.plot(cum_returns.index, cum_returns[col], label=col, linewidth=2)
    
    ax2.set_title('Cumulative Returns (6M)', fontsize=12, fontweight='bold')
    ax2.set_ylabel('Cumulative Return')
    ax2.legend(fontsize=8)
    ax2.grid(True, alpha=0.3)
    plt.xticks(rotation=45)

# Plot 3: Asset Allocation
ax3 = plt.subplot(2, 3, 3)
allocation_query = """
SELECT Asset_Type, COUNT(*) as Count
FROM Strategy_Log
WHERE Backtest_Run_ID = (
    SELECT Backtest_Run_ID FROM Strategy_Log 
    ORDER BY Run_Date DESC LIMIT 1
)
AND Portfolio_Rank IS NOT NULL
GROUP BY Asset_Type
"""
alloc_data = db.execute_query_dict(allocation_query)
if alloc_data:
    df_alloc = pd.DataFrame(alloc_data)
    colors = ['#e74c3c', '#3498db', '#2ecc71', '#f39c12']
    ax3.pie(df_alloc['Count'], labels=df_alloc['Asset_Type'], autopct='%1.1f%%',
           colors=colors, textprops={'fontsize': 9})
    ax3.set_title('Latest Portfolio Allocation', fontsize=12, fontweight='bold')

# Plot 4: ETF Performance (Latest Portfolio)
ax4 = plt.subplot(2, 3, 4)
perf_query = """
SELECT Ticker_Symbol, Momentum_Score
FROM Strategy_Log
WHERE Backtest_Run_ID = (
    SELECT Backtest_Run_ID FROM Strategy_Log 
    ORDER BY Run_Date DESC LIMIT 1
)
AND Portfolio_Rank IS NOT NULL
ORDER BY Portfolio_Rank
LIMIT 5
"""
perf_data = db.execute_query_dict(perf_query)
if perf_data:
    df_perf = pd.DataFrame(perf_data)
    ax4.barh(df_perf['Ticker_Symbol'], df_perf['Momentum_Score'], color='steelblue')
    ax4.set_title('Top 5 ETFs by Momentum', fontsize=12, fontweight='bold')
    ax4.set_xlabel('Momentum Score (%)')
    ax4.grid(axis='x', alpha=0.3)

# Plot 5: Weekly Volatility Comparison
if vol_results:
    ax5 = plt.subplot(2, 3, 5)
    ax5.bar(df_vol_plot['Asset_Type'], df_vol_plot['Weekly_Volatility_Pct'],
           color='coral')
    ax5.set_title('Weekly Volatility', fontsize=12, fontweight='bold')
    ax5.set_ylabel('Volatility (%)')
    ax5.grid(axis='y', alpha=0.3)
    plt.xticks(rotation=45)

# Plot 6: Backtest Performance Summary
ax6 = plt.subplot(2, 3, 6)
metrics = ['Cum Return', 'CAGR', 'Rebalances']
values = [
    results['cumulative_return']*100,
    results['cagr']*100,
    results['total_rebalances']*10  # Scale for visibility
]
colors_bar = ['#2ecc71', '#3498db', '#9b59b6']
ax6.bar(metrics, values, color=colors_bar)
ax6.set_title('Backtest Performance', fontsize=12, fontweight='bold')
ax6.set_ylabel('Value')
ax6.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✅ All visualizations generated!")

# ========== FINAL SUMMARY ==========
print("\n" + "="*80)
print("🎉 DEMO COMPLETE - SYSTEM SUMMARY")
print("="*80)

print("\n✅ **What Was Demonstrated:**")
print("   1️⃣  Integrated System Architecture (Main Controller + 5 Sub-modules)")
print("   2️⃣  Momentum Backtesting Engine (Real Yahoo Finance Data)")
print("   3️⃣  SQL-Based Analytics (3 Complex Insights)")
print("   4️⃣  CRUD Operations (Database Management)")
print("   5️⃣  Interactive Visualizations (6 Charts)")

print("\n📊 **Key Results:**")
print(f"   • ETFs Analyzed: {etf_count}")
print(f"   • Price Records: {price_count:,}")
print(f"   • Backtest Period: {results['total_rebalances']} rebalances")
print(f"   • Cumulative Return: {results['cumulative_return']*100:.2f}%")
print(f"   • Annualized CAGR: {results['cagr']*100:.2f}%")

print("\n🏗️  **System Architecture:**")
print("   Main Controller (ETFBacktesterSystem)")
print("   ├── DatabaseConnector (MySQL)")
print("   ├── MomentumBacktester (Strategy Engine)")
print("   ├── PortfolioAnalytics (SQL Insights)")
print("   ├── CRUDOperations (Database Management)")
print("   └── DataLoader (Yahoo Finance)")

print("\n✨ **All Fixes Applied (20 Nov 2025):**")
print("   ✅ Fixed import statements")
print("   ✅ Fixed method names")
print("   ✅ Fixed backtest parameters")
print("   ✅ Fixed Decimal/float conversion")
print("   ✅ Fixed SQL ONLY_FULL_GROUP_BY compatibility")

# Clean shutdown
db.close_pool()

print("\n" + "="*80)
print("💚 Thank you! - DADS 4002 Course Project")
print("="*80)
```
