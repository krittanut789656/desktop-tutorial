## 4️⃣ Analytics - วิเคราะห์ Portfolio

### 4.1 Insight 1: Risk-Adjusted Performance Analysis

````python
# วิเคราะห์ความเสี่ยงของ portfolios 1, 2, 3
result = etf.run_risk_analysis(
    portfolio_ids=[1, 2, 3],
    start_date='2020-01-01',
    end_date='2024-12-31',
    benchmark='SPY',
    db_config=DB_CONFIG
)
````

### 4.2 Insight 2: Optimal Rebalancing Frequency

````python
# วิเคราะห์ความถี่ rebalancing ที่ดีที่สุด
# ⚠️ จะใช้เวลา 3-5 นาที (run 5 backtests)
result = etf.run_rebalancing_analysis(
    portfolio_id=1,
    start_date='2020-01-01',
    end_date='2024-12-31',
    initial_capital=100000.0,
    db_config=DB_CONFIG
)
````

````python
# แสดงผลลัพธ์สรุป
if result['success']:
    print(f"🏆 Best Strategy: {result['best_strategy']}")
    print(f"📈 Best Return: {result['best_return']:.2f}%")
    print(f"📊 Best Sharpe: {result['best_sharpe']:.3f}")
    print(f"\n💡 {result['recommendation']}")
````

````python
# อ่าน report เต็ม
with open('insight2_rebalancing_analysis.txt', 'r', encoding='utf-8') as f:
    print(f.read())
````

### 4.3 Insight 3: DCA vs Lump Sum

````python
# เปรียบเทียบ DCA vs Lump Sum
result = etf.run_dca_analysis(
    portfolio_id=1,
    total_capital=100000.0,
    investment_months=12,
    start_date='2020-01-01',
    end_date='2024-12-31',
    db_config=DB_CONFIG
)
````

````python
# แสดงผลลัพธ์
if result['success']:
    print(f"💰 Lump Sum Return: {result['lumpsum_annualized_return']:.2f}%")
    print(f"📊 DCA Return: {result['dca_annualized_return']:.2f}%")
    print(f"🏆 Winner: {result['winner']}")
    print(f"📈 DCA Win Rate: {result['dca_win_rate']:.1f}%")
````

````python
# อ่าน report เต็ม
with open('insight3_dca_vs_lumpsum.txt', 'r', encoding='utf-8') as f:
    print(f.read())
````

## 7️⃣ Tips & Advanced Usage

### ใช้ Pandas วิเคราะห์ต่อ

````python
# ดู portfolios แบบกรองตาม category
etfs_df = etf.show_etfs(DB_CONFIG)

# กรองเฉพาะ US Equity
us_equity = etfs_df[etfs_df['category'] == 'US Equity']
display(us_equity)
````

````python
# วิเคราะห์ backtest performance
backtests_df = etf.show_backtest_history(limit=50, db_config=DB_CONFIG)

# Portfolio ไหนให้ return ดีที่สุด
best_backtests = backtests_df.nlargest(5, 'total_return')
display(best_backtests[['portfolio_name', 'strategy_type', 'total_return']])
````

### Custom Queries

````python
# Query ข้อมูลเอง
import mysql.connector

conn = mysql.connector.connect(**DB_CONFIG)

query = """
SELECT ticker, AVG(close) as avg_price, COUNT(*) as records
FROM daily_prices
WHERE date >= '2024-01-01'
GROUP BY ticker
ORDER BY avg_price DESC
LIMIT 10
"""

df = pd.read_sql(query, conn)
conn.close()

display(df)
````
